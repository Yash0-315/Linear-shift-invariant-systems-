```python
import numpy as np
import matplotlib.pyplot as plt

# Define discrete time index n
n = np.arange(-5, 16)

# Input signal: 4-point rectangular pulse x[n] = u[n] - u[n-4]
x = np.where((n >= 0) & (n <= 3), 1.0, 0.0)

# Impulse response: causal exponential decay h[n] = (0.7)^n * u[n]
h = np.where(n >= 0, (0.7)**n, 0.0)

# Compute linear convolution y[n] = x[n] * h[n]
y_full = np.convolve(x, h)
n_full = np.arange(n[0] + n[0], n[-1] + n[-1] + 1)
mask = (n_full >= -5) & (n_full <= 15)
y = y_full[mask]

# Verify Shift-Invariance (shift input by n0 = 4)
n0 = 4
x_shifted = np.where((n >= n0) & (n <= 3 + n0), 1.0, 0.0)
y_from_shifted = np.convolve(x_shifted, h)[mask]
y_shifted_direct = np.roll(y, n0)
y_shifted_direct[:n0] = 0

# Verify Linearity (a1*x1 + a2*x2)
a1, a2 = 2.0, -1.5
x2 = np.where(n == 1, 1.0, 0.0)
y2 = np.convolve(x2, h)[mask]
y_combined_in = np.convolve(a1 * x + a2 * x2, h)[mask]
y_combined_out = a1 * y + a2 * y2

# Numerical Verification Output
print("Max Shift-Invariance Error:", np.max(np.abs(y_from_shifted - y_shifted_direct)))
print("Max Linearity Error:", np.max(np.abs(y_combined_in - y_combined_out)))
print("BIBO Stability Sum |h[n]|:", round(np.sum(np.abs(h)), 4))

# Plotting Input, Impulse Response, and Convolved Output
fig, axes = plt.subplots(3, 1, figsize=(9, 8.5))
axes[0].stem(n, x, linefmt='C0-', markerfmt='C0o', basefmt='k-')
axes[0].set_title("Input Signal x[n] = u[n] - u[n-4]")
axes[1].stem(n, h, linefmt='C1-', markerfmt='C1s', basefmt='k-')
axes[1].set_title("Impulse Response h[n] = (0.7)^n u[n]")
axes[2].stem(n, y, linefmt='C2-', markerfmt='C2D', basefmt='k-')
axes[2].set_title("Output Response y[n] = x[n] * h[n]")
for ax in axes:
    ax.set_ylabel("Amplitude")
    ax.grid(True)
axes[-1].set_xlabel("Discrete Time Index (n)")
plt.tight_layout()
plt.show()
```
